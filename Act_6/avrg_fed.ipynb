```python
def fedavg(models):

    model_weights = [m.get_weights() for m in models]

    avg_weights = []

    for weights in zip(*model_weights):

        avg_weights.append(
            np.mean(weights, axis=0)
        )

    return avg_weights
```

```python
def weighted_fedavg(models, client_sizes):

    total = sum(client_sizes)

    model_weights = [
        m.get_weights()
        for m in models
    ]

    global_weights = []

    for weights in zip(*model_weights):

        weighted = np.zeros_like(weights[0])

        for w, n in zip(weights, client_sizes):

            weighted += w * (n / total)

        global_weights.append(weighted)

    return global_weights
```

```python
def median_aggregation(models):

    model_weights = [
        m.get_weights()
        for m in models
    ]

    global_weights = []

    for weights in zip(*model_weights):

        global_weights.append(
            np.median(
                np.array(weights),
                axis=0
            )
        )

    return global_weights
```

```python
def build_global_fedavg(local_models):
    
    global_model = build.build_model()

    weights = fedavg(local_models)

    global_model.set_weights(weights)

    return global_model
```
