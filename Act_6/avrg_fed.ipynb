### FedAvg

```python
def fedavg(models):

    model_weights = [m.get_weights() for m in models]

    avg_weights = []

    for weights in zip(*model_weights):

        avg_weights.append(
            np.mean(weights, axis=0)
        )

    return avg_weights
```

### Trimmed mean

https://arxiv.org/pdf/1803.01498

```python
def trimmed_mean(models, trim_ratio=0.25):

    model_weights = [
        m.get_weights()
        for m in models
    ]

    global_weights = []

    for weights in zip(*model_weights):
        w = np.array(weights)
        sorted_w = np.sort(w, axis=0)
        k = int(len(w) * trim_ratio)
        trimmed = sorted_w[k:len(w)-k]


        global_weights.append(
            np.mean(trimmed, axis=0)
        )

    return global_weights
```

### FedMedian

https://arxiv.org/pdf/1912.13445

```python
def fedmedian(models):

    model_weights = [
        m.get_weights()
        for m in models
    ]

    global_weights = []

    for weights in zip(*model_weights):

        global_weights.append(
            np.median(
                np.array(weights),
                axis=0
            )
        )

    return global_weights
```

## Comparison

```python
import os

for f in os.listdir("."):
    if f.endswith(".keras"):
        print(f)
```

```python
fedavg_model = tf.keras.models.load_model("global_fedavg.keras")

fedmedian_model = tf.keras.models.load_model("global_fedmedian.keras")

trimmed_model = tf.keras.models.load_model("global_trimmed_mean.keras")
```

```python
import tensorflow as tf
import numpy as np

train, test = tf.keras.datasets.mnist.load_data()

x_train, x_test = np.expand_dims(train[0] / 255.0, -1), np.expand_dims(test[0] / 255.0, -1)
y_train, y_test = train[1], test[1]
```

```python
from sklearn.metrics import classification_report
import numpy as np

for name, model in [
    ("FedAvg", fedavg_model),
    ("FedMedian", fedmedian_model),
    ("TrimmedMean", trimmed_model)
]:

    y_pred = np.argmax(
        model.predict(x_test),
        axis=1
    )

    print("\n", name)
    print(
        classification_report(
            y_test,
            y_pred
        )
    )
```

```python
fedavg_model.evaluate(x_test, y_test)

fedmedian_model.evaluate(x_test, y_test)

trimmed_model.evaluate(x_test, y_test)
```
